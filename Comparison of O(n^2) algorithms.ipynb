```python
#Selection sort implementation
# Array size = 1000
import time
import copy  # To make copies of the arrays

# Selection Sort function
def selection_sort(list_a):
    for i in range(len(list_a) - 1):
        min_index = i  # Assume the current element is the smallest

        for j in range(i + 1, len(list_a)):
            if list_a[j] < list_a[min_index]:
                min_index = j  # Update the index of the smallest element

        if min_index != i:
            list_a[i], list_a[min_index] = list_a[min_index], list_a[i]  # Swap values

# ----------------------------- SELECTION SORT TIMING -----------------------------
selection_sort_times = []

for index, original_array in enumerate(original_bubble_arrays_1000, 1):
    arr_copy = copy.deepcopy(original_array)  # Copy to preserve the original array

    start_time = time.time()
    selection_sort(arr_copy)
    end_time = time.time()

    time_taken = end_time - start_time
    selection_sort_times.append(time_taken)
    print(f"Time taken for selection sort on array {index}: {time_taken:.6f} seconds")

# Calculate and print average selection sort time
avg_selection_sort_time = sum(selection_sort_times) / len(selection_sort_times)
print(f"\nAverage time taken for selection sort: {avg_selection_sort_time:.6f} seconds")
```
