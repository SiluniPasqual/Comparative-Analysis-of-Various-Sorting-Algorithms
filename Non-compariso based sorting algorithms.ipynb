```python
#Radix sort implementation
#Array size = 1000
import time
import copy  # To make copies of the arrays

# Counting Sort used in Radix Sort
def counting_sort(l, place):
    size = len(l)
    output = [0] * size
    count = [0] * 10

    for i in range(size):
        index = (l[i] // place) % 10
        count[index] += 1
    
    for j in range(1, 10):
        count[j] += count[j-1]

    k = size - 1
    while k >= 0:
        index = (l[k] // place) % 10
        output[count[index] - 1] = l[k]
        count[index] -= 1
        k -= 1

    for a in range(size):
        l[a] = output[a]

# Radix Sort function
def radix_sort(l):
    max_val = max(l)
    place = 1
    while max_val // place > 0:
        counting_sort(l, place)
        place *= 10

# ----------------------------- REUSING COUNTING SORT ARRAYS -----------------------------
original_radix_sort_arrays_1000 = copy.deepcopy(original_counting_sort_arrays_1000)

# ----------------------------- RADIX SORT TIMING -----------------------------
radix_sort_times = []

for index, original_array in enumerate(original_radix_sort_arrays_1000, 1):
    arr_copy = copy.deepcopy(original_array)  # Copy to preserve the original array

    start_time = time.time()
    radix_sort(arr_copy)
    end_time = time.time()

    time_taken = end_time - start_time
    radix_sort_times.append(time_taken)
    print(f"Time taken for radix sort on array {index}: {time_taken:.6f} seconds")

# Calculate and print average radix sort time
avg_radix_sort_time = sum(radix_sort_times) / len(radix_sort_times)
print(f"\nAverage time taken for radix sort (Array Size = 1000): {avg_radix_sort_time:.6f} seconds")
```
