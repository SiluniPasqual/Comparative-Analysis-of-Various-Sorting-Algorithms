```python
#Comparison of O(nlogn) algorithms

#Quicksort algorithm implementation
#Arraysize = 1000
import random
import time
import copy  # To make copies of the arrays

# Partition function
def partition(arr, low, high):
    pivot_index = random.randint(low, high)  # Select a random pivot
    pivot = arr[pivot_index]

    arr[pivot_index], arr[high] = arr[high], arr[pivot_index]  # Swap pivot with last element

    i = low - 1  # Pointer for the smaller element
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

# Quick Sort function
def randomizedQuickSort(arr, low, high):
    if low < high:
        pivot_index = partition(arr, low, high)
        randomizedQuickSort(arr, low, pivot_index - 1)
        randomizedQuickSort(arr, pivot_index + 1, high)

# ----------------------------- GENERATING ARRAYS -----------------------------
num_arrays = 3
array_size = 1000
original_quick_sort_arrays_1000 = [[random.randint(1, 1000) for _ in range(array_size)] for _ in range(num_arrays)]

# ----------------------------- QUICK SORT TIMING -----------------------------
quick_sort_times = []

for index, original_array in enumerate(original_quick_sort_arrays_1000, 1):
    arr_copy = copy.deepcopy(original_array)  # Copy to preserve the original array

    start_time = time.time()
    randomizedQuickSort(arr_copy, 0, len(arr_copy) - 1)
    end_time = time.time()

    time_taken = end_time - start_time
    quick_sort_times.append(time_taken)
    print(f"Time taken for quick sort on array {index}: {time_taken:.6f} seconds")

# Calculate and print average quick sort time
avg_quick_sort_time = sum(quick_sort_times) / len(quick_sort_times)
print(f"\nAverage time taken for quick sort: {avg_quick_sort_time:.6f} seconds")
```

```python
#Mergesort implementation
#Arraysize = 1000
import time
import copy  # To make copies of the arrays

# Merge function
def merge(arr, left, mid, right):
    left_part = arr[left:mid + 1]  # Left subarray
    right_part = arr[mid + 1:right + 1]  # Right subarray

    i, j, k = 0, 0, left  # Initialize pointers

    while i < len(left_part) and j < len(right_part):
        if left_part[i] <= right_part[j]:
            arr[k] = left_part[i]
            i += 1
        else:
            arr[k] = right_part[j]
            j += 1
        k += 1

    # Copy remaining elements (if any)
    while i < len(left_part):
        arr[k] = left_part[i]
        i += 1
        k += 1

    while j < len(right_part):
        arr[k] = right_part[j]
        j += 1
        k += 1

# Merge Sort function
def mergeSort(arr, left, right):
    if left < right:
        mid = (left + right) // 2
        mergeSort(arr, left, mid)
        mergeSort(arr, mid + 1, right)
        merge(arr, left, mid, right)

# ----------------------------- MERGE SORT TIMING -----------------------------
merge_sort_times = []

for index, original_array in enumerate(original_quick_sort_arrays_1000, 1):  # Using the same 3 arrays from Quick Sort
    arr_copy = copy.deepcopy(original_array)  # Copy to preserve the original array

    start_time = time.time()
    mergeSort(arr_copy, 0, len(arr_copy) - 1)
    end_time = time.time()

    time_taken = end_time - start_time
    merge_sort_times.append(time_taken)
    print(f"Time taken for merge sort on array {index}: {time_taken:.6f} seconds")

# Calculate and print average merge sort time
avg_merge_sort_time = sum(merge_sort_times) / len(merge_sort_times)
print(f"\nAverage time taken for merge sort: {avg_merge_sort_time:.6f} seconds")
```

```python
#Heapsort implementation
#Arraysize = 1000
import time
import copy  # To make copies of the arrays

# Swap function
def swap(lst, i, j):
    lst[i], lst[j] = lst[j], lst[i]

# Sift down function for heapify
def siftDown(lst, i, upper):
    while True:
        l, r = 2 * i + 1, 2 * i + 2  # Left and right child indices

        if max(l, r) < upper:
            if lst[i] >= max(lst[l], lst[r]):
                break
            elif lst[l] > lst[r]:
                swap(lst, i, l)
                i = l
            else:
                swap(lst, i, r)
                i = r

        elif l < upper:
            if lst[l] > lst[i]:
                swap(lst, i, l)
                i = l
            else:
                break
        elif r < upper:
            if lst[r] > lst[i]:
                swap(lst, i, r)
                i = r
            else:
                break
        else:
            break

# Heap Sort function
def heapsort(lst):
    # Build max heap
    for j in range((len(lst) - 2) // 2, -1, -1):
        siftDown(lst, j, len(lst))

    # Extract elements from heap
    for end in range(len(lst) - 1, 0, -1):
        swap(lst, 0, end)
        siftDown(lst, 0, end)

# ----------------------------- HEAP SORT TIMING -----------------------------
heap_sort_times = []

# Using the same 3 arrays from QuickSort (original_quick_sort_arrays_1000)
for index, original_array in enumerate(original_quick_sort_arrays_1000, 1):
    arr_copy = copy.deepcopy(original_array)  # Copy to preserve the original array

    start_time = time.time()
    heapsort(arr_copy)
    end_time = time.time()

    time_taken = end_time - start_time
    heap_sort_times.append(time_taken)
    print(f"Time taken for heap sort on array {index}: {time_taken:.6f} seconds")

# Calculate and print average heap sort time
avg_heap_sort_time = sum(heap_sort_times) / len(heap_sort_times)
print(f"\nAverage time taken for heap sort: {avg_heap_sort_time:.6f} seconds")
```
